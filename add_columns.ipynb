```python
import pandas as pd
import numpy as np
df = pd.read_excel("clean_coast_data.xlsx")
```

```python
#sameeee coast number
df.loc[(df.coast_name == 'חוף כזיב') | (df.coast_name == 'כזיב')]['coast_num'].unique()
df.loc[(df.coast_name == 'חוף בצת') | (df.coast_name == 'בצת')]['coast_num'].unique()

#change the name to one of em
df.loc[df['coast_name'] == 'כזיב', 'coast_name'] = 'חוף כזיב'
df.loc[df['coast_name'] == 'בצת', 'coast_name'] = 'חוף בצת'
```

add distance from port ashdod

```python
ashdod_port = (31.82803555650259, 34.65294228673902)
#we have 2 beaches with the same name, one is in eilat the other in tel aviv apperantly
lighthouse_center_coast = (32.10333081350449, 34.77694988873057)
```

```python
from geopy.distance import geodesic


cords = pd.read_excel("coordinates.xlsx")


# Merge the DataFrames on 'coast_name'
merged_df = df.merge(cords, left_on='coast_name',right_on='coast_name')
# merged_df
merged_df.loc[((merged_df['coast_name'] == 'חוף המגדלור') & (merged_df.Area == 'Center')), 'lat'] = lighthouse_center_coast[0]
merged_df.loc[((merged_df['coast_name'] == 'חוף המגדלור') & (merged_df.Area == 'Center')), 'lon'] = lighthouse_center_coast[1]
```

```python
merged_df
```

```python
def get_distance(row):
    lat= row['lat']
    lon= row['lon']
    return geodesic((lat,lon),ashdod_port).km

merged_df['distance_to_ashdod_port'] = merged_df.apply(lambda row: get_distance(row), axis=1)
```

```python
merged_df.drop(['Unnamed: 0_y', 'lat', 'lon','Unnamed: 0_x'], axis = 1, inplace=True)
```
