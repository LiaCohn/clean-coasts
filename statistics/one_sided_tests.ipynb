```python
import pandas as pd
import scipy.stats as stats
import numpy as np # for linear algebra operations
import plotly.graph_objects as go # for visualization
df = pd.read_excel("../clean_coast_data.xlsx")
```

```python
df.head()
```

```python
df['test_result'].unique()
```

```python
cols_to_test = ['bathing_season', 'summer_vaction','winter','nature','river','military']
for col in cols_to_test:
    #list(df.loc[(df['Tel_Aviv-Yafo_pm2.5_exceeded'] == 1) & (df['region_Central']==1)]['Total'])
    negetive = df.loc[(df[col] == 0) & (df['test_result'] <= 3)][['test_result']]
    positive = df.loc[(df[col] == 1) & (df['test_result'] <= 3)][['test_result']]
    stat, p = stats.mannwhitneyu(negetive.test_result, positive.test_result,alternative='greater')

    print('stat=%.3f, p=%.3f' % (stat, p))
    if p > 0.05:
        print('Probably the same distribution')
    else:
        print('Probably different distributions')
```
