```python
import pandas as pd
import scipy.stats as stats
df = pd.read_excel("../clean_coast_data.xlsx")
```

```python
import numpy as np # for linear algebra operations
import plotly.graph_objects as go # for visualization
```

```python
corr_df = df.corr()
fig = go.Figure()
fig.add_traces(go.Heatmap(
    z=corr_df, 
    x= corr_df.columns, 
    y= corr_df.columns,
))
fig.update_layout({
    'title':"Coasts Correlation Heatmap",

})
fig.show()
```

```python
for index in corr_df.index:
  corr_values = []
  for col in corr_df.columns:
    if index!=col and abs(corr_df[index][col])>0.85:
      corr_values.append(col)
  if len(corr_values) > 0:
    print(index+ ' is correlated to: \n' , corr_values)
```

```python
test_df = df.dropna()
columns_to_test = ['city rank','Distance from gaza','hifa_port_dis']
target_col = 'test_result'
for col in columns_to_test:
    # # Calculate the Spearman rank correlation between two columns
    rho, p_value = stats.spearmanr(test_df[target_col], test_df[col])
    print("Spearman rank correlation:", round(rho,3), round(p_value,5))
```

```python
dates = test_df['Date test'].view(np.int64)
rho, p_value = stats.spearmanr(test_df[target_col], dates)
print("Spearman rank correlation:", round(rho,3), round(p_value,5))
```

```python
summer_vac_res = test_df.loc[test_df['summer_vaction'] == 1][['test_result','Date test']]
winter_res = test_df.loc[test_df['winter'] == 1][['test_result','Date test']]
eilat_res = df.loc[df['Area'] == 'Eilat'][['test_result','Date test']]
subsets = [summer_vac_res, winter_res , eilat_res]
for subset in subsets:
    subset['Date test'] = subset['Date test'].view(np.int64)
    rho, p_value = stats.spearmanr(subset.test_result, subset['Date test'])
    print("Spearman rank correlation:", round(rho,3), round(p_value,5))
```
