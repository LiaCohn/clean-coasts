```python
import pandas as pd
import scipy.stats as stats
df = pd.read_excel("../clean_coast_data.xlsx")
```

```python
import numpy as np # for linear algebra operations
import plotly.graph_objects as go # for visualization
```

```python
corr_df = df.corr()
fig = go.Figure()
fig.add_traces(go.Heatmap(
    z=corr_df, 
    x= corr_df.columns, 
    y= corr_df.columns,
))
fig.update_layout({
    'title':"Coasts Correlation Heatmap",

})
fig.show()
```

```python
for index in corr_df.index:
  corr_values = []
  for col in corr_df.columns:
    if index!=col and abs(corr_df[index][col])>0.85:
      corr_values.append(col)
  if len(corr_values) > 0:
    print(index+ ' is correlated to: \n' , corr_values)
```

```python
df.drop(['Unnamed: 0'], axis=1, inplace=True)
df.columns
```

```python
#get data from 2016 onwards.
df_2016 = df.loc[df.year >= 2016]
df_2016_eilat = df_2016.loc[df_2016.Area == 'Eilat'] #just eilat coasts
df_2016_rest = df_2016.loc[df_2016.Area != 'Eilat']
df_2016_eilat
```

```python
# test_df = df.dropna()
columns_to_test = ['city rank','Distance from gaza','hifa_port_dis','distance_to_ashdod_port']
target_col = 'test_result_num'
for col in columns_to_test:
    # # Calculate the Spearman rank correlation between two columns
    rho, p_value = stats.spearmanr(df_2016_rest[target_col], df_2016_rest[col])
    print("Spearman rank correlation:", round(rho,3), round(p_value,5))

# print()
#There is no variation in df_2016_eilat['city rank'] so its standard deviation is equal 
#to 0 which will result in zero division in the spearmanr() function, thereby returning a NaN.
# rho, p_value = stats.spearmanr(df_2016_eilat[target_col], df_2016_eilat['city rank'])
# print("Spearman rank correlation:", round(rho,3), round(p_value,5))
```

```python
dates = df_2016_rest['Date test'].view(np.int64)
rho, p_value = stats.spearmanr(df_2016_rest[target_col], dates)
print("Spearman rank correlation all sites:", round(rho,3), round(p_value,5))
dates = df_2016_eilat['Date test'].view(np.int64)
rho, p_value = stats.spearmanr(df_2016_eilat[target_col], dates)
print("Spearman rank correlation Eilat sites:", round(rho,3), round(p_value,5))
```

```python
print("Correlation using all sites")
summer_vac_res = df_2016_rest.loc[df_2016_rest['summer_vaction'] == 1][['test_result_num','Date test']]
winter_res = df_2016_rest.loc[df_2016_rest['winter'] == 1][['test_result_num','Date test']]
# eilat_res = df.loc[df['Area'] == 'Eilat'][['test_result','Date test']]
subsets = [summer_vac_res, winter_res ]
for subset in subsets:
    subset['Date test'] = subset['Date test'].view(np.int64)
    rho, p_value = stats.spearmanr(subset.test_result_num, subset['Date test'])
    print("Spearman rank correlation:", round(rho,3), round(p_value,5))

print()
print("Correlation using Eilat sites")

summer_vac_res = df_2016_eilat.loc[df_2016_eilat['summer_vaction'] == 1][['test_result_num','Date test']]
winter_res = df_2016_eilat.loc[df_2016_eilat['winter'] == 1][['test_result_num','Date test']]
# eilat_res = df.loc[df['Area'] == 'Eilat'][['test_result','Date test']]
subsets = [summer_vac_res, winter_res ]
for subset in subsets:
    subset['Date test'] = subset['Date test'].view(np.int64)
    rho, p_value = stats.spearmanr(subset.test_result_num, subset['Date test'])
    print("Spearman rank correlation:", round(rho,3), round(p_value,5))
```
