```python
import pandas as pd
import scipy.stats as stats
import numpy as np # for linear algebra operations
import plotly.graph_objects as go # for visualization
df = pd.read_excel("../clean_coast_data.xlsx")
```

```python
cols_to_test = ['bathing_season', 'summer_vaction','winter','nature','river','military','have_rained']
for col in cols_to_test:
    negative = df.loc[(df[col] == 0)][['test_result_num']].dropna()
    positive = df.loc[(df[col] == 1)][['test_result_num']].dropna()
    stat, p = stats.mannwhitneyu(positive.test_result_num, negative.test_result_num,alternative='two-sided')
    std_0,mean_0,median_0 = negative.std(),negative.mean(),negative.median()
    std_1,mean_1,median_1 = positive.std(),positive.mean(),positive.median()

    print("positive group values: std=%.3f, mean=%.3f, median=%.4f" %(std_1, mean_1,median_1))
    print("negative group values: std=%.3f, mean=%.3f, median=%.4f" %(std_0, mean_0,median_0))

    print('stat=%.3f, p=%.5f' % (stat, p))
    if p > 0.05:
        print(f"""{col}: Probably the same distribution"""+'\n')
    else:
        print(f"""{col}: Probably different distributions"""+'\n')
```

```python
areas = df['Area'].unique()
areas_dict = {x:list(df.loc[(df['Area'] == x)]['test_result_num']) for x in areas}
[print(f"{area}: std=%.3f, mean=%.3f, median=%.4f" 
       %(np.std(areas_dict[area]), np.mean(areas_dict[area]),np.median(areas_dict[area]))) for area in areas_dict.keys()]

for i in range(len(areas) - 1):
    for j in range(i+1, len(areas)):
        if areas[i] != areas[j]:
            area = areas_dict[areas[i]]
            area_to_compare_to = areas_dict[areas[j]]

            stat, p = stats.mannwhitneyu(area, area_to_compare_to,alternative='two-sided')

            std_0,mean_0,median_0 = np.std(area), np.mean(area),np.median(area)
            std_1,mean_1,median_1 = np.std(area_to_compare_to), np.mean(area_to_compare_to),np.median(area_to_compare_to)


            print('stat=%.3f, p=%.8f' % (stat, p))
            if p > 0.05:
                print(f"""{areas[i]} vs {areas[j]}: Probably the same distribution"""+'\n')
            else:
                print(f"""{areas[i]} vs {areas[j]}: Probably different distributions"""+'\n')
```

```python
merged_df = df.merge(authorities, left_on='coast_name',right_on='coast_name')
```

```python
import pandas as pd
from scipy import stats

# Assuming df is your DataFrame

# Add 'local_authority' to the cols_to_test list
cols_to_test = ['bathing_season', 'summer_vaction', 'winter', 'nature', 'river', 'military', 'rain']

# Iterate through unique local_authority values
for local_authority_value in merged_df['local_authority'].unique():
    print(f"Local Authority: {local_authority_value}")
    
    # Filter DataFrame for the specific local_authority
    local_authority_df = merged_df[merged_df['local_authority'] == local_authority_value]

    # Iterate through the columns for statistical analysis
    for col in cols_to_test:  # Skip the 'local_authority' column
        if col in local_authority_df.columns:
            negative = local_authority_df.loc[(local_authority_df[col] == 0)][['test_result_num']].dropna()
            positive = local_authority_df.loc[(local_authority_df[col] == 1)][['test_result_num']].dropna()

            # Perform the Mann-Whitney U test
            stat, p = stats.mannwhitneyu(positive.test_result_num, negative.test_result_num, alternative='two-sided')

            # Calculate statistics for both groups
            std_0, mean_0, median_0 = negative.std(), negative.mean(), negative.median()
            std_1, mean_1, median_1 = positive.std(), positive.mean(), positive.median()

            # Print the results
            print(f"{col}:")
            print("Positive group values - std=%.3f, mean=%.3f, median=%.4f" % (std_1, mean_1, median_1))
            print("Negative group values - std=%.3f, mean=%.3f, median=%.4f" % (std_0, mean_0, median_0))
            print('stat=%.3f, p=%.5f' % (stat, p))

            # Check for significance and print the result
            if p > 0.05:
                print("Probably the same distribution\n")
            else:
                print("Probably different distributions\n")
```
